目前想要结合 TRO-Grasp / Get-zero / hora / DexNDM 来走手型-物体的双向泛化叙事

TRO-Grasp 的特色是 interaction-centric 的框架叙事，用的属于MPNN网络，有OR/RR注意力，边的特征很强，但它不属于策略网络，而是 grasp synthesis 合成

Get-zero 的特色是 zero-shot embodiment generalization，用的是 Graph Transformer，属于策略网络，但它的边特征很弱，且只是 robot-embodiment graph。同时它无法泛化到leaphand family以外的手构型，无物体泛化能力

hora 的特色是 rapid adaptation module, 通过本体感受形成对物体属性推测的隐状态，从而构成物体的泛化

dexndm 的特色是 joint-wize 的因式分解关节动力学，对sim2real gap的缩小有很大帮助。如果做物体泛化，可以考虑 joint-wize/finger-wize的rapid adaptation module的隐变量

有一些未形成系统性，但局部认为有效的构思：

1. urdf等特征一定也要作为输入，并和关节状态做cross-attention
2. 如果要实现物体泛化，

当下我觉得，如果一口气去做物体-手双向泛化，有点过于贪婪，仅能实现手型的手内操作任务真正泛化，便已能超越get-zero并产出一篇成果了，而物体-手型双向泛化，可以放到后续的工作中去实现，作为第二篇成果。

如果是这样的话，边界进一步缩减。我们主要围绕Get-zero和TRO-Grasp展开工作
